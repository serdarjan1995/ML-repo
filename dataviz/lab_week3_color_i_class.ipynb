```python
# create a HISTOGRAM, which shows distribution of data, with the BODY MASS data for 3 species
# on the y-axis, a count of the # of penguins
# on the x-axis, a measure of body mass (g)
# each species should have its own COLOR - choose wisely!
# experiment with many colors until you find some that work well to you
# how do the colors look next to each other? (Albers)
# make sure to add a legend and a title to the histogram

unique_penquins = penguins.Species.unique()
adelie = penguins[penguins.Species == unique_penquins[0]]
gentoo = penguins[penguins.Species == unique_penquins[1]]
chinstrap = penguins[penguins.Species == unique_penquins[2]]


fig = plt.figure()

plt.hist(adelie["Body Mass (g)"], color="steelblue", ls="dotted", edgecolor='gray', lw=1, alpha = 0.7, label="Adelie")
plt.hist(gentoo["Body Mass (g)"], color="green", ls="dashed", edgecolor='gray', alpha = 0.7, label="Gentoo")
plt.hist(chinstrap["Body Mass (g)"], color="wheat", ls="solid", edgecolor='gray', alpha = 0.7, label="Chinstrap")
plt.xlabel("Body mass (g)")
plt.ylabel("Number of Penguins")
plt.legend()
plt.title("Penguins and their mass histogram")

plt.plot()














```
